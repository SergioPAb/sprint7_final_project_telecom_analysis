## 🧩Paso 4: Summary statistics de uso por usuario


### 4.1 Agrupación por comportamiento de uso

🎯**Objetivo**: Resumir las variables clave de la tabla `usage` **por usuario**, creando métricas que representen su comportamiento real de uso histórico.

**Instrucciones:**:
1. Construye una tabla agregada de `usage` por `user_id` que incluya:
- número total de mensajes  
- número total de llamadas  
- total de minutos de llamadas

2. Renombra las columnas para que tengan nombres claros:  
- `cant_mensajes`  
- `cant_llamadas`  
- `cant_minutos_llamada`
3. Combina esta tabla con `users`.

```python

# Columnas auxiliares
usage["is_text"] = (usage["type"] == "text").astype(int)
usage["is_call"] = (usage["type"] == "call").astype(int)

# Agrupar información por usuario
usage_agg = usage.groupby('user_id').agg(
    is_text=('is_text', 'sum'),
    is_call=('is_call', 'sum'),
    duration=('duration', 'sum')
).reset_index()

# observar resultado
print(usage_agg.head(3))
```

```python

# Renombrar columnas
usage_agg = usage_agg.rename(columns={
    'is_text': 'cant_mensajes',
    'is_call': 'cant_llamadas',
    'duration': 'cant_minutos_llamada'
})

# observar resultado
print(usage_agg.head(3))
```

```python


# Combinar la tabla agregada con el dataset de usuarios

user_profile = users.merge(usage_agg, on='user_id', how='left')
user_profile.head(5)
```

  - Si solamente hay outliers de un solo lado, no es necesario calcular ambos límites.

**Hint:**
- Dentro del bucle, usa `plt.title(f'Boxplot: {col}')` para que el título cambie acorde a la columna.

```python
# Visualizando usando BoxPlot
columnas_numericas = ['age', 'cant_mensajes', 'cant_llamadas', 'cant_minutos_llamada']

for col in columnas_numericas:
    plt.figure(figsize=(8, 4))
    sns.boxplot(data=user_profile, x=col, palette=['skyblue'])
    plt.title(f'Boxplot: {col}')
    plt.xlabel(col)
    plt.show()
```

💡Insights:
- Age: El boxplot de edad muestra una distribución simétrica, con la mediana alrededor de los 35 años. No se observan puntos fuera de los bigotes (whiskers), por lo que no presenta outliers. Los valores están contenidos dentro del rango esperado (aproximadamente 18-70 años).
  
- cant_mensajes: El boxplot de cantidad de mensajes muestra una distribución con una gran cantidad de puntos por encima del bigote superior, lo que indica la presencia de múltiples outliers en el lado superior. La mayoría de los usuarios se concentran en rangos bajos de mensajes (0-5), pero existen usuarios con volúmenes de mensajes excepcionalmente altos que superan los 10-15 mensajes.

- cant_llamadas: El boxplot de cantidad de llamadas muestra una distribución con una gran cantidad de puntos por encima del bigote superior, lo que indica la presencia de múltiples outliers en el lado superior. La mayoría de los usuarios se concentran en rangos bajos de llamadas (0-4), pero existen usuarios con volúmenes de llamadas excepcionalmente altos que superan las 8-10 llamadas.

- cant_minutos_llamada: El boxplot de minutos de llamada muestra una distribución con una gran cantidad de puntos por encima del bigote superior, lo que indica la presencia de múltiples outliers en el lado superior. La mayoría de los usuarios se concentran en rangos bajos de minutos (0-20), pero existen usuarios con volúmenes de minutos excepcionalmente altos que superan los 60-80 minutos.

```python

# Calcular límites con el método IQR
columnas_limites = ['cant_mensajes', 'cant_llamadas', 'cant_minutos_llamada']

for col in columnas_limites:
    Q1 = user_profile[col].quantile(0.25)
    Q3 = user_profile[col].quantile(0.75)
    IQR = Q3 - Q1
    limite_superior = Q3 + 1.5 * IQR

    print(f"--- {col} ---")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Límite superior: {limite_superior:.2f}")
    print(f"Outliers superiores: {(user_profile[col] > limite_superior).sum()}")
    print()
```

```python

# Revisa los limites superiores y el max, para tomar la decisión de mantener los outliers o no
user_profile[columnas_limites].describe()
```

💡Insights:
- cant_mensajes: El límite superior se encuentra en 11.50 mensajes, mientras que el valor máximo registrado es de 17.00 mensajes, con 46 usuarios (1.15% del total) identificados como outliers. Aunque existe esta diferencia, los outliers representan un porcentaje muy bajo del total de usuarios. Estos usuarios con alto volumen de mensajes pueden representar un segmento de alto valor que merece ser analizado por separado, por lo que se decide mantener los outliers.

- cant_llamadas: El límite superior se encuentra en 10.50 llamadas, mientras que el valor máximo registrado es de 15.00 llamadas, con 30 usuarios (0.75% del total) identificados como outliers. El porcentaje es mínimo y el máximo no es excesivamente lejano al límite superior. Al igual que con los mensajes, estos usuarios con alto volumen de llamadas pueden corresponder a clientes con necesidades reales de mayor consumo, por lo que se decide mantener los outliers.

🎯 **Objetivo:** Clasificar a cada usuario en un grupo de uso (Bajo uso, Uso medio, Alto uso) basándose en la cantidad de llamadas y mensajes registrados.

**Instrucciones:**  
- Crea una nueva columna llamada `grupo_uso` en el dataframe `user_profile`.
- Usa comparaciones lógicas (<, >) para evaluar las condiciones de llamadas y mensajes y asigna:
  - `'Bajo uso'` cuando llamadas < 5 y mensajes < 5
  - `'Uso medio'` cuando llamadas < 10 y mensajes < 10
  - `'Alto uso'` para el resto de casos

```python
# Crear columna grupo_uso

user_profile['grupo_uso'] = 'Alto uso'  # valor por defecto para el resto de casos

user_profile.loc[(user_profile['cant_llamadas'] < 5) & (user_profile['cant_mensajes'] < 5), 'grupo_uso'] = 'Bajo uso'
user_profile.loc[(user_profile['cant_llamadas'] < 10) & (user_profile['cant_mensajes'] < 10), 'grupo_uso'] = 'Uso medio'
```

### 6.2 Segmentación de Clientes Por Edad

🎯 **Objetivo:**: Clasificar a cada usuario en un grupo por **edad**.

**Instrucciones:**  
- Crea una nueva columna llamada `grupo_edad` en el dataframe `user_profile`.
- Usa comparaciones lógicas (<, >) para evaluar las condiciones y asigna:
  - `'Joven'` cuando age < 30
  - `'Adulto'` cuando age < 60
  - `'Adulto Mayor'` para el resto de casos

```python
# Crear columna grupo_edad
condiciones = [
    user_profile['age'] < 30,
    user_profile['age'] < 60
]

resultados = ['Joven', 'Adulto']

user_profile['grupo_edad'] = np.select(condiciones, resultados, default='Adulto Mayor')
```
